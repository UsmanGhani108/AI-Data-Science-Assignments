```python
# Ques 07 (Write a program to take two numbers and an operator (/,x,+,-) as input and perform the corresponding operation.)
ques07_no1 = int(input("Enter first number : "))
ques07_no2 = int(input("Enter second number : "))
operator = input("Choose operators (+ , - , * , /)")
if operator == "+" :
    print("Addition = " , ques07_no1 + ques07_no2)
elif operator == "-" :
    print("Subtraction = " , ques07_no1 - ques07_no2)
elif operator == "*" :
    print("Multiplication = " , ques07_no1 * ques07_no2)
elif operator == "/" :
    print("Division = " , ques07_no1 / ques07_no2)
else:
    print("Choose Right Operators")
```

```python
# Ques 11 (Write a program that takes a temperature in Celsius and checks if it is freezing, moderate or hot.
# a. Freezing temperature is below 0.
# b. Moderate temperature is greater than 0 and less than 26.
# c. Hot temperature is above 26. )
ques11_temp = int(input("Enter temperature : "))
if ques11_temp < 0 :
    print("Freezing Temperature")
elif ques11_temp > 0 and ques11_temp < 26:
    print("Moderate temperature")
else:
    print("Hot Temperature")
```

```python
# Ques 12 (Write a program to input salary, if salary is greater than Rs. 600000 then deduct 5% tax and 
# calculate the Gross Salary.)
ques12_salary = int(input("Enter your salary : "))
if ques12_salary > 600000 :
    tax_amount = ques12_salary * 0.05
    gross_salary = ques12_salary - tax_amount
    print(f"Total Tax amount : {tax_amount}")
    print(f"Gross Salary amount : {gross_salary}")    
else:
    print("Tax amount is zero")
    print(f"Gross Salary {ques12_salary}") 
```
