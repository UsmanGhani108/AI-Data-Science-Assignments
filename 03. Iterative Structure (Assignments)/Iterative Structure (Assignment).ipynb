```python
# Solution : 02 --> While Loop
num = int(input("Enter any  number : "))
factorial = 1
i = 1
while i <= num:
    factorial *= i
    i += 1
print(f"Factorial : {factorial}")
```

# Question 11

```python
# Solution : 01 --> For Loop
even_sum = 0
odd_sum = 0
for i in range(1,51):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i
print(f"Total Even Sum : {even_sum}")
print(f"Total odd Sum : {odd_sum}")
```

```python
# Solution : 02 --> While Loop
even_sum = 0 
odd_sum = 0
i = 1
while i <=50:
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i
    i += 1
print(f"Total Even Sum : {even_sum}")
print(f"Total odd Sum : {odd_sum}")
```

# Question 14

```python
number_of_guests = 300
discard_menu = ["fish" , "qeema" , "nihari"]
menu = []
menu_price = {}
for i in range(5):
    food_item = input("Enter dish name : ")
    if food_item.lower() in discard_menu:
        print(f"Sorry {food_item} is not allowed")
        continue
    menu.append(food_item)
    price = int(input(f"Enter price of {food_item} : "))
    menu_price[food_item] = price
print(f"Menu : {menu}")
print(f"Menu Price : {menu_price}")
total_menu_price = 0
for price in menu_price.values():
    total_menu_price = total_menu_price + price
print(f"Total Menu Price : {total_menu_price}")
menu_price_perhead = {}
for food_item , price in menu_price.items():
    perhead_cost = price / number_of_guests 
    menu_price_perhead[food_item] = perhead_cost
print(f"Per Head Prices : {menu_price_perhead}")
complete_perhead_cost = 0
for price in menu_price_perhead.values():
    complete_perhead_cost += price
print("Complete Menu Per Head Cost:", complete_perhead_cost)
```
