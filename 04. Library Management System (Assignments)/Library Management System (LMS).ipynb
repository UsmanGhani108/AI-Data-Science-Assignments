# Library Management System (LMS)

## The functionalities of this project
1. Add Book
2. View All Books
3. Search Book
4. Update Book
5. Delete Book
6. Issue Book
7. Return Book
8. Exit

- List of Dictionaries
- Structure of data storing books: books = [{Book01}, {Book02}, {Book03}]
- Each book stores its ID, title, author, total stock, and available copies.

```python
books = []
keys = ["Book ID", "Title", "Author", "Total Copies", "Available Copies"]
book = dict.fromkeys(keys, '')
```

```python
# Defining Add Book Function:
def add_book():
    print("------------------------")
    book_id = input("Enter Book ID : ")

    for b in books:
        if b["Book ID"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title : ")
    author = input("Enter Author Name : ")
    copies = int(input("Enter Number of Copies : "))

    bk = book.copy()
    bk.update(
        {
            "Book ID": book_id,
            "Title": title,
            "Author": author,
            "Total Copies": copies,
            "Available Copies": copies
        }
    )

    books.append(bk)
    print("Book Added Successfully!!!!")
    print("------------------------")
    return
```

```python
# Defining Retrieve Books Function:
def display_books():
    if not books:
        print("Books are empty!")
    else:
        for index, book in enumerate(books):
            print("--------------------------------")
            print(f"Book {index+1}")
            for key, value in book.items():
                print(f"{key} : {value}")
            print("--------------------------------")
```

```python
# Defining Search Book Function:
def search_book(book_id):
    if not books:
        print("Unable to search in empty books list")
    else:
        print("--------------------------------")
        for book in books:
            if book["Book ID"] == book_id:
                for key, value in book.items():
                    print(f"{key} : {value}")
                return
        print("Book not found!")
        print("--------------------------------")
```

```python
# Defining Update Book Function:
def update_book(search_book_id):
    if not books:
        print("Unable to update empty books list")
    else:
        for index, book in enumerate(books):
            if book["Book ID"] == search_book_id:
                updated_title = input("Enter title to update : ")
                updated_author = input("Enter author to update : ")

                books[index].update(
                    {
                        "Title": updated_title,
                        "Author": updated_author
                    }
                )

                print("Book Updated Successfully!!!!")
                return

        print("Book not found!")
```

```python
# Defining Delete Book Function:
def delete_book(search_book_id):
    if not books:
        print("Unable to delete from empty books list")
    else:
        for book in books:
            if book["Book ID"] == search_book_id:
                books.remove(book)
                print("Remove successfully.")
                return

        print("Book not found!")
```

```python
# Defining Issue Book Function:
def issue_book(search_book_id):
    if not books:
        print("Unable to issue from empty books list")
    else:
        for book in books:
            if book["Book ID"] == search_book_id:
                if book["Available Copies"] > 0:
                    book["Available Copies"] -= 1
                    print("Book Issued Successfully!!!!")
                else:
                    print("Book is not available.")
                return

        print("Book not found!")
```

```python
# Defining Return Book Function:
def return_book(search_book_id):
    if not books:
        print("Unable to return book in empty books list")
    else:
        for book in books:
            if book["Book ID"] == search_book_id:
                if book["Available Copies"] < book["Total Copies"]:
                    book["Available Copies"] += 1
                    print("Book Returned Successfully!!!!")
                else:
                    print("No issued copies of this book to return.")
                return

        print("Book not found!")
```

```python
# Menu driven program
while True:
    print("------------- LMS -------------")
    print("1. Press {c} to add book")
    print("2. Press {r} to retrieve books")
    print("3. Press {s} to search book")
    print("4. Press {u} to update book")
    print("5. Press {d} to delete book")
    print("6. Press {i} to issue book")
    print("7. Press {t} to return book")
    print("8. Press {e} to exit")

    option = input("Enter your choice : ").lower()

    if option == "c":
        # Add Book record
        add_book()

    elif option == "r":
        # Retrieve all book records
        display_books()

    elif option == "s":
        # Search book record by ID
        search_book_id = input("Enter Book ID to search : ")
        search_book(search_book_id)

    elif option == "u":
        # Update book record by ID
        search_book_id = input("Enter Book ID to update : ")
        update_book(search_book_id)

    elif option == "d":
        # Delete book record by ID
        search_book_id = input("Enter Book ID to delete : ")
        delete_book(search_book_id)

    elif option == "i":
        # Issue book by ID
        search_book_id = input("Enter Book ID to issue : ")
        issue_book(search_book_id)

    elif option == "t":
        # Return book by ID
        search_book_id = input("Enter Book ID to return : ")
        return_book(search_book_id)

    elif option == "e":
        break

    else:
        print("Invalid Choice")
```
